```python
import numpy as np

# Parameters
gamma = 0.9
states = [0, 1, 2, 3, 4]
actions = ["L", "R"]

# Deterministic transition function
def step(s, a):
    if s == 4:
        return 4, 0  # terminal
    if a == "R":
        s_next = min(s + 1, 4)
    else:  # "L"
        s_next = max(s - 1, 0)
    reward = 1 if s_next == 4 else 0
    return s_next, reward

# Initialize Q arbitrarily
Q = {(s, a): 0.0 for s in states for a in actions}

# Value iteration on Q*
for iteration in range(1000):
    Q_new = Q.copy()
    delta = 0.0

    for s in states:
        for a in actions:
            s_next, r = step(s, a)
            if s == 4:
                target = 0.0
            else:
                target = r + gamma * max(Q[(s_next, a2)] for a2 in actions)
            Q_new[(s, a)] = target
            delta = max(delta, abs(Q_new[(s, a)] - Q[(s, a)]))

    Q = Q_new
    if delta < 1e-10:
        break

# Print results
print("Q* values:")
for s in states:
    for a in actions:
        print(f"Q({s}, {a}) = {Q[(s, a)]:.6f}")
    print()

print("Derived optimal policy:")
for s in states:
    best_action = max(actions, key=lambda a: Q[(s, a)])
    print(f"pi({s}) = {best_action}")
```

```python
import numpy as np

gamma = 0.9
states = [0,1,2,3,4]
actions = ["L","R"]

def transitions(s, a):
    if s == 4:
        return [(4, 0, 1.0)]  # terminal

    if a == "L":
        s_next = max(s-1, 0)
        r = 1 if s_next == 4 else 0
        return [(s_next, r, 1.0)]

    if a == "R":
        s_right = min(s+1, 4)
        s_left  = max(s-1, 0)

        r_right = 1 if s_right == 4 else 0
        r_left  = 1 if s_left == 4 else 0

        return [
            (s_right, r_right, 0.8),
            (s_left,  r_left,  0.2)
        ]

Q = {(s,a): 0.0 for s in states for a in actions}

for iteration in range(1000):
    Q_new = Q.copy()
    delta = 0.0

    for s in states:
        for a in actions:
            total = 0.0
            for s_next, r, p in transitions(s,a):
                total += p * (r + gamma * max(Q[(s_next,a2)] for a2 in actions))
            Q_new[(s,a)] = total
            delta = max(delta, abs(Q_new[(s,a)] - Q[(s,a)]))

    Q = Q_new
    if delta < 1e-10:
        break

print("Q* values:")
for s in states:
    for a in actions:
        print(f"Q({s}, {a}) = {Q[(s,a)]:.6f}")
    print()

print("Optimal policy:")
for s in states:
    best = max(actions, key=lambda a: Q[(s,a)])
    print(f"pi({s}) = {best}")
```
